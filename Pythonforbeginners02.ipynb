```python
# BMI using function
Name1 = "M"
Height_m1 = 1.7
Weight_kg1 = 62

Name2 = "F"
Height_m2 = 1.2
Weight_kg2 = 50

Name3 = "K"
Height_m3 = 1.8
Weight_kg3 = 85

def bmi_calculator(name, height_m, weight_kg):
    bmi = weight_kg / (height_m ** 2)
    bmi = round(bmi, 2)
    print("bmi: ")
    print(bmi)
    print(name)
    if bmi < 25:
        return name + " is not over weight"
    else:
        return name + " is over weight"
result1 = bmi_calculator(Name1, Height_m1, Weight_kg1)
result2 = bmi_calculator(Name2, Height_m2, Weight_kg2)
result3 = bmi_calculator(Name3, Height_m3, Weight_kg3)
print(result1)
print(result2)
print(result3)
```

```python
def km_to_miles(kilometers):
  """Converts kilometers to miles.

  Args:
    kilometers: The distance in kilometers.

  Returns:
    The distance in miles.
  """
  miles = kilometers * 0.621371  # 1 kilometer is approximately 0.621371 miles
  miles = round(miles, 2)
  return miles

convert = km_to_miles(100)
print(convert)
```

```python
def miles_to_km(miles):
  """Converts miles to kilometers.

  Args:
    miles: The distance in miles.

  Returns:
    The distance in kilmoeters.
  """
  kilometers = miles * 1.60934  # 1 mile is approximately 1.60934 kilometers
  kilometers = round(kilometers, 2)
  return kilometers

convert = miles_to_km(72)
print(convert)
```

```python
print(list(range(1, 100)))
def get_multiples_of_3_and_5():
  """Gets a list of numbers between 1 and 99 that are multiples of 3 or 5.

  Returns:
    A list of numbers that are multiples of 3 or 5.
  """
  numbers = list(range(1, 100))
  multiples = []
  for number in numbers:
    if number % 3 == 0 or number % 5 == 0:
      multiples.append(number)
  return multiples

multiples_of_3_and_5 = get_multiples_of_3_and_5()
print(multiples_of_3_and_5)
```
